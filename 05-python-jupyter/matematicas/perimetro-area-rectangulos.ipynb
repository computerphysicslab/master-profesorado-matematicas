## 1. Pregunta matemática

Un huerto rectangular debe tener **área** $A = 24\,\mathrm{m}^2$.

## 3. Modelo

- Área: $A = a \cdot b$ (m²).  
- Perímetro: $P = 2(a + b)$ (m).  
- Si $A$ es fija y $a > 0$, entonces $b = A / a$ y  
  $$P(a) = 2\left(a + \frac{A}{a}\right).$$

Supuesto: rectángulo plano, lados positivos; en la tabla usamos lados enteros.

```python
def area(a, b):
    return a * b

def perimetro(a, b):
    return 2 * (a + b)

# Contraejemplo rápido: misma área, distinto perímetro
ejemplos = [(6, 4), (12, 2), (8, 3), (24, 1)]
print(f"{'a':>4} {'b':>4} {'A':>6} {'P':>6}")
for a, b in ejemplos:
    print(f"{a:4} {b:4} {area(a, b):6} {perimetro(a, b):6}")

print("\nSi A es igual y P no, la afirmación «misma área ⇒ mismo perímetro» es FALSA.")
```

## 4. Tabla: lados enteros con $A = 24$

Listamos divisores de 24 (lado $a$ en metros) y el perímetro asociado.

```python
A_fija = 24
# Divisores positivos de 24 (lado a); b = 24/a entero
pares = []
for a in range(1, A_fija + 1):
    if A_fija % a == 0:
        b = A_fija // a
        # Evitar duplicar (3,8) y (8,3) si quieres solo a <= b:
        if a <= b:
            pares.append((a, b, perimetro(a, b)))

print(f"{'a':>4} {'b':>4} {'P (m)':>8}")
for a, b, P in pares:
    print(f"{a:4} {b:4} {P:8}")

P_min = min(pares, key=lambda t: t[2])
P_max = max(pares, key=lambda t: t[2])
print(f"\nMenor perímetro: {P_min[0]}×{P_min[1]} → P = {P_min[2]} m")
print(f"Mayor perímetro: {P_max[0]}×{P_max[1]} → P = {P_max[2]} m")
```

## 5. ¿Cabe con 22 m de valla?

```python
presupuesto = 22  # metros de valla
factibles = [(a, b, P) for a, b, P in pares if P <= presupuesto]
if factibles:
    print("Sí es posible. Opciones con P ≤ 22 m:")
    for a, b, P in factibles:
        print(f"  {a}×{b} → P = {P} m")
else:
    print("No hay rectángulo de lados enteros con A = 24 y P ≤ 22.")
    print("El menor P de la tabla es:", P_min[2], "m")
```

## 6. Gráfica: $P$ en función del lado $a$ (área fija)

Aunque $a$ no sea entero, con $b = 24/a$ el perímetro cambia con la forma.

```python
import matplotlib.pyplot as plt
import numpy as np

A = 24
a_vals = np.linspace(0.5, 24, 200)
P_vals = 2 * (a_vals + A / a_vals)

plt.figure(figsize=(7, 4))
plt.plot(a_vals, P_vals, label=r"$P(a) = 2(a + 24/a)$")
# Marcar pares enteros a <= b
for a, b, P in pares:
    plt.scatter([a], [P], zorder=5)
    plt.annotate(f"{a}×{b}", (a, P), textcoords="offset points", xytext=(4, 4), fontsize=8)
plt.xlabel("Lado a (m)")
plt.ylabel("Perímetro P (m)")
plt.title("Área fija A = 24 m²: el perímetro depende de la forma")
plt.grid(True, alpha=0.3)
plt.legend()
plt.tight_layout()
plt.show()
```
