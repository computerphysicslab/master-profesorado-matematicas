## 1. Pregunta matemática

Dos tarifas de datos:

| Tarifa | Cuota fija (€/mes) | Coste por MB extra (€) |
|--------|-------------------|-------------------------|
| **A**  | 10                | 0,05                    |
| **B**  | 6                 | 0,12                    |

1. ¿A partir de qué consumo mensual $x$ (MB extra) conviene cada una?  
2. ¿Qué ocurre si la cuota de A sube 2 €?  
3. Un anuncio dice que B es «más barata». ¿Para quién es verdad?

Modelo:  
$$y_A = 0{,}05\, x + 10,\quad y_B = 0{,}12\, x + 6$$

## 3. Modelo en código

Parámetros con nombres claros. Unidades: $x$ en MB; $y$ en €.

```python
# Coste mensual en función del consumo extra x (MB)
def coste_A(x, cuota=10.0, por_mb=0.05):
    return por_mb * x + cuota

def coste_B(x, cuota=6.0, por_mb=0.12):
    return por_mb * x + cuota

# Punto de corte: 0.05x + 10 = 0.12x + 6  →  4 = 0.07x
x_corte = (10 - 6) / (0.12 - 0.05)
print(f"Punto de corte: x ≈ {x_corte:.2f} MB")
print(f"Coste en el corte: {coste_A(x_corte):.2f} €")
```

## 4. Tabla de valores

Compara con una tabla hecha a mano (0, 50, 100, 150 MB).

```python
xs = [0, 50, 80, 100, 150]
print(f"{'x (MB)':>8}  {'A (€)':>8}  {'B (€)':>8}  {'mejor':>6}")
for x in xs:
    a, b = coste_A(x), coste_B(x)
    mejor = "A" if a < b else ("B" if b < a else "empate")
    print(f"{x:8}  {a:8.2f}  {b:8.2f}  {mejor:>6}")
```

## 5. Gráfica de las dos tarifas

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(0, 200, 200)
plt.figure(figsize=(7, 4))
plt.plot(x, coste_A(x), label="A: 0,05x + 10")
plt.plot(x, coste_B(x), label="B: 0,12x + 6")
plt.axvline(x_corte, color="gray", linestyle="--", alpha=0.7, label=f"corte ≈ {x_corte:.1f} MB")
plt.xlabel("Consumo extra x (MB)")
plt.ylabel("Coste mensual y (€)")
plt.title("Comparación de tarifas")
plt.legend()
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```
