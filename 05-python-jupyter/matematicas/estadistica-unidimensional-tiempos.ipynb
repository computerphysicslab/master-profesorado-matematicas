## 3. Modelo (qué calculamos)

Sea una lista de $n$ datos ordenados $x_{(1)} \leq x_{(2)} \leq \cdots \leq x_{(n)}$.

| Medida | Definición (síntesis) |
|--------|------------------------|
| **Media** $\bar x$ | $\bar x = \dfrac{1}{n}\sum_{i=1}^{n} x_i$ |
| **Mediana** $Me$ | Valor central si $n$ impar; media de los dos centrales si $n$ par |
| **Rango** | $\max - \min$ |
| **Moda** | Valor más frecuente (puede haber más de una) |

**Variable:** tiempo de resolución (minutos), cuantitativa continua en origen, aquí registrada en enteros.  
**Población/muestra:** 10 cronometrajes ficticios de un mismo problema.

```python
import statistics as stats

# Datos ficticios (minutos). No sustituir por notas reales de alumnos.
tiempos = [8, 9, 7, 10, 8, 12, 9, 11, 8, 45]

n = len(tiempos)
media = stats.mean(tiempos)
mediana = stats.median(tiempos)
rango = max(tiempos) - min(tiempos)
try:
    moda = stats.mode(tiempos)  # unimodal; si hay empate, statistics.mode toma uno
except stats.StatisticsError:
    moda = None

print(f"n = {n}")
print(f"Media   = {media:.2f} min")
print(f"Mediana = {mediana:.2f} min")
print(f"Rango   = {rango} min")
print(f"Moda    = {moda} min")
print(f"Datos ordenados: {sorted(tiempos)}")
```

## 4. ¿Qué pasa si quitamos el 45?

Comparamos el mismo resumen **con** y **sin** el valor atípico.

```python
sin_atipico = [t for t in tiempos if t != 45]

def resumen(nombre, datos):
    print(f"--- {nombre} (n={len(datos)}) ---")
    print(f"  Media   = {stats.mean(datos):.2f}")
    print(f"  Mediana = {stats.median(datos):.2f}")
    print(f"  Rango   = {max(datos) - min(datos)}")

resumen("Con el 45", tiempos)
resumen("Sin el 45", sin_atipico)

print("\nObserva cuánto se mueve la media y cuánto la mediana.")
```

## 5. Gráfica: distribución de los tiempos

Un diagrama de barras (o histograma) hace visible el **45** respecto al resto del grupo.

```python
import matplotlib.pyplot as plt

fig, ax = plt.subplots(figsize=(7, 4))
ax.hist(tiempos, bins=[6, 8, 10, 12, 14, 46], edgecolor="black", align="mid")
ax.axvline(stats.mean(tiempos), color="C1", linestyle="--", label=f"Media = {stats.mean(tiempos):.1f}")
ax.axvline(stats.median(tiempos), color="C2", linestyle="-", label=f"Mediana = {stats.median(tiempos):.1f}")
ax.set_xlabel("Tiempo (min)")
ax.set_ylabel("Frecuencia")
ax.set_title("Tiempos de resolución (datos ficticios)")
ax.legend()
ax.grid(True, axis="y", alpha=0.3)
plt.tight_layout()
plt.show()
```

## 6. Variante rápida (opcional)
